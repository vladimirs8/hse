```python
import requests

# Отправляем GET-запрос к API
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)

# Проверяем успешность запроса
if response.status_code == 200:
    posts = response.json()

    # Выводим первые 5 постов
    for post in posts[:5]:
        print("Заголовок:", post["title"])
        print("Тело:", post["body"])
        print("-" * 50)
else:
    print("Ошибка запроса:", response.status_code)
```

```python
import requests
from getpass import getpass

# Вводим API-ключ безопасно
api_key = getpass("Введите ваш OpenWeather API key: ")

# Запрашиваем название города
city = input("Введите название города: ")

# Адрес API
url = "https://api.openweathermap.org/data/2.5/weather"

# Параметры запроса
params = {
    "q": city,
    "appid": api_key,
    "units": "metric",
    "lang": "ru"
}

# Отправляем GET-запрос
response = requests.get(url, params=params)

# Обрабатываем ответ
if response.status_code == 200:
    weather = response.json()

    temperature = weather["main"]["temp"]
    description = weather["weather"][0]["description"]

    print(f"\nПогода в городе {city}:")
    print(f"Температура: {temperature} °C")
    print(f"Описание: {description}")

elif response.status_code == 401:
    print("Ошибка: неверный или ещё не активированный API-ключ.")

elif response.status_code == 404:
    print("Ошибка: город не найден.")

else:
    print("Ошибка запроса:", response.status_code)
    print(response.text)
```
